```python
import pandas as pd
```

```python
apple = pd.read_csv('archive/stocks/AAPL.csv')
facebook = pd.read_csv('archive/stocks/FB.csv')
google = pd.read_csv('archive/stocks/GOOGL.csv')
nvidia = pd.read_csv('archive/stocks/NVDA.csv')
tesla = pd.read_csv('archive/stocks/TSLA.csv')
twitter = pd.read_csv('archive/stocks/TWTR.csv')
```

```python
dfs = [apple, facebook, google, nvidia, tesla, twitter]
```

```python
# Apply moving average
for df in dfs:
    df['MA50'] = df.Close.rolling(50).mean()
    df['MA200'] = df.Close.rolling(200).mean()
```

```python
apple
```

```python
for df in dfs:
    df['Previous Day close price'] = df.Close.shift(1)
```

```python
apple
```

```python
for df in dfs:
    df['Change in price'] = df['Close'] - df['Previous Day close price']
```

```python
apple
```

```python
for df in dfs:
    df['Percent change in price'] = df.Close.pct_change()
```

```python
apple
```

```python
for df in dfs:
    df['Previous day volume'] = df.Volume.shift(1)
```

```python
apple
```

```python
for df in dfs:
    df['Change in volume'] = df['Volume'] - df['Previous day volume']
```

```python
apple
```

```python
for df in dfs:
    df['Percent change in volume'] = df.Volume.pct_change()
```
